```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
clusterA = np.array(
    [np.random.beta(a = 2, b = 8, size = 100),
    np.random.beta(a = 3, b = 9, size = 100)]
)

clusterB = np.array(
    [np.random.beta(a = 8, b = 2, size = 100),
    np.random.beta(a = 9, b = 3, size = 100)]
)


clusterC = np.array(
    [np.random.beta(a = 10, b = 3, size = 100),
    np.random.beta(a = 11, b = 13, size = 100)]
)
data = np.array([clusterA, clusterB, clusterC])
# Data has to have the shape (n_clusters, 2, n_samples)

plt.figure(figsize=(10,10))
plt.xlim(0,1)
plt.ylim(0,1)
for x, y in data:
    plt.scatter(x, y)

```
